3. Operasi MAtriks dan Vektor

```python
# Penjumlahan Matriks
import numpy as np

A = np.array([[80,30,25],
              [50,20,10]])

B = np.array([[30,60,15],
              [10,30,30]])

hasil = A + B

print(hasil)
```

```python
#Dot Product
import numpy as np

A = np.array([10,20,15])
B = np.array([16,67,35])

hasil = np.dot(A,B)

print(hasil)
```

```python
# Cross Product
import numpy as np

A = np.array([10,20,15])
B = np.array([16,67,35])

hasil = np.cross(A,B)

print(hasil)
```

**TUGAS PARAKTIKUM**

```python
# 1. Perkalian Matriks
import numpy as np

A = np.array([[3,1],
              [2,0]])

B = np.array([[2,1],
              [1,-1]])

C = np.array([[-1,1,2],
              [2,0,-1]])

# A x B
hasil_AB = np.dot(A,B)

# B . C
hasil_BC = np.dot(B,C)

print("Hasil A x B =")
print(hasil_AB)

print("\nHasil B . C =")
print(hasil_BC)
```

```python
# 2. Faktorial
import math

# 6!
f1 = math.factorial(6)

# 8! / 5!
f2 = math.factorial(8) / math.factorial(5)

print("6! =", f1)
print("8! / 5! =", f2)
```

```python
# 3. Operasi Vektor
import numpy as np

A = np.array([5,3,7])
B = np.array([12,-3,1])

# Panjang vektor
panjang_A = np.linalg.norm(A)
panjang_B = np.linalg.norm(B)

# Dot product
dot_AB = np.dot(A,B)

# Cross product
cross_AB = np.cross(A,B)

print("Panjang Vektor A =", panjang_A)
print("Panjang Vektor B =", panjang_B)

print("A . B =", dot_AB)

print("A x B =", cross_AB)
```
